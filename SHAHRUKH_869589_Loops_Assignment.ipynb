## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python

balance = 0
account_created = False


print('''
____________________________________________________________________________________________

                            Banking System
____________________________________________________________________________________________



        1. Create Account

        2. Deposit

        3. Withdraw 

        4. Check Balance

        5. Exit
''')


for i in range(100):

    choice = input("Enter your choice: ")

    if choice == "1":
        if account_created == False:
            name = input("Enter your name: ")
            account_created = True
            print("Account created successfully!")
        else:
            print("Account already exists.")

    elif choice == "2":
        if account_created == False:
            print("Please create an account first.")
            continue

        amount = int(input("Enter deposit amount: "))

        if amount <= 0:
            print("Invalid deposit amount.")
            continue

        balance = balance + amount
        print("Amount deposited successfully!")

    elif choice == "3":
        if account_created == False:
            print("Please create an account first.")
            continue

        amount = int(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance = balance - amount
            print("Withdrawal successful!")

    elif choice == "4":
        if account_created == False:
            print("Please create an account first.")
        else:
            print("Your balance is:", balance)

    elif choice == "5":
        print("Thank you for using our banking system!")
        break

    else:
        print("Invalid choice. Please try again.")
```

## Create a quiz game using loops and conditions.

```python
all_questions = []

print('''
__________________________________________________________________________________________

                        Quiz Application In Python
________________________________________________________________________________________

    1 Press : Add New Question 
    
    2 Press : Start Quiz
    
    3 Press : Delete Question
    
    4 Press : Update Question
    
    5 Press : Quit
''')

for i in range(100):

    choice = int(input('Enter the choice: '))

    # ADD QUESTION
    if choice == 1:

        new_question = input(
            f'Enter the new question no {len(all_questions) + 1}: '
        )

        choices = []

        for i in range(4):
            choice = input(f'Enter the choice no {i + 1}: ')
            choices.append(choice)

        correct_answer = input('Enter the correct answer: ')

        all_questions.append({
            'question': new_question,
            'choices': choices,
            'correct': correct_answer
        })

        print(
            f'Question no {len(all_questions)} added successfully'
        )

    # START QUIZ
    elif choice == 2:

        print('''
_____________________________________________________________________________________

                            Start Your Quiz
_____________________________________________________________________________________
''')

        if len(all_questions) == 0:
            print('No questions available.')
            continue

        score = 0

        for index, question in enumerate(all_questions):

            print(f"Q{index + 1}: {question['question']}")

            for index, choice in enumerate(question['choices']):
                print(f'{index + 1}) {choice}')

            user_answer = input('Enter the correct answer: ')

            actual_answer = question['correct']

            if user_answer == actual_answer:
                score += 1
                print('Correct Answer')

            else:
                print(
                    f'Wrong Answer! Correct Answer is {actual_answer}'
                )

            print(
                '________________________________________________________________________'
            )

        print(f'''
Total Questions : {len(all_questions)}

Correct Answers : {score}

Wrong Answers   : {len(all_questions) - score}
''')

    # DELETE QUESTION
    elif choice == 3:

        delete_question_no = int(
            input('Enter the question no: ')
        )

        total_questions = len(all_questions)

        if delete_question_no <= total_questions and delete_question_no > 0:

            del all_questions[delete_question_no - 1]

            print(
                f'Question No {delete_question_no} deleted successfully'
            )

        else:
            print(
                f'Question No {delete_question_no} not found'
            )

    # UPDATE QUESTION
    elif choice == 4:

        update_question_no = int(
            input('Enter the question no: ')
        )

        total_questions = len(all_questions)

        if update_question_no <= total_questions and update_question_no > 0:

            question = all_questions[update_question_no - 1]

            new_question = input('Enter new question: ')

            choices = []

            for i in range(4):
                new_choice = input(
                    f'Enter new choice {i + 1}: '
                )
                choices.append(new_choice)

            correct_answer = input(
                'Enter new correct answer: '
            )

            question['question'] = new_question
            question['choices'] = choices
            question['correct'] = correct_answer

            print(
                f'Question No {update_question_no} updated successfully'
            )

        else:
            print(
                f'Question No {update_question_no} not found'
            )

    # QUIT
    elif choice == 5:

        print('Thanks for using this quiz application')
        break

    else:
        print('Invalid choice')
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []

for day in range(1, 6):
    total = 0

    print(f"Day {day}:")

    for product in range(1, 4):
        sale = int(input(f"Product {product}: "))

        if sale == 0:
            print("Skipped")
            continue

        total = total + sale

    daily_totals.append(total)

    print("Daily Total:", total)

print("Daily Totals =", daily_totals)
print("Grand Total =", sum(daily_totals))
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
total_booked = 0
total_available = 0

for floor in range(1, 4):
    booked = 0
    available = 0

    print(f"\nFloor {floor}:")

    for room in range(1, 5):
        status = input(f"Room {room} (Y/N): ").upper()

        if status == "Y":
            booked += 1
            total_booked += 1

        else:
            available += 1
            total_available += 1

    print(f"Floor {floor} = {booked} Booked, {available} Available")

print("\nHotel Booking Summary")
print("Total Booked Rooms:", total_booked)
print("Total Available Rooms:", total_available)
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []

for invoice in range(1, 4):
    items = []

    print(f"\nInvoice {invoice}:")

    for item in range(1, 5):
        price = int(input(f"Enter Item {item} Price: "))
        items.append(price)

    subtotal = sum(items)
    tax = subtotal * 10 / 100
    total = subtotal + tax

    invoice_totals.append(total)

    print("Items:", items)
    print("Subtotal:", subtotal)
    print("Tax (10%):", tax)
    print("Invoice Total:", total)

print("\nAll Invoice Totals:", invoice_totals)
```

```python
for student in range(1, 6):
    marks_list = []
    failed = False

    print(f"\nStudent {student}:")

    for subject in range(1, 5):
        marks = int(input(f"Subject {subject} Marks: "))
        marks_list.append(marks)

        if marks < 33:
            print(f"Failed in Subject {subject}")
            failed = True

    total = sum(marks_list)
    percentage = total / 400 * 100

    if failed == True:
        grade = "Fail"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "E"

    print("Marks:", marks_list)
    print("Total Marks:", total)
    print("Percentage:", percentage)
    print("Grade:", grade)
    
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ["Hotel", "Transport", "Food"]

for package in range(1, 4):
    ratings = []

    print(f"\nPackage {package}:")

    for feature in features:
        rating = int(input(f"Enter {feature} Rating (0-5): "))

        if rating == 0:
            print("Rating Skipped")
            continue

        ratings.append(rating)

    if len(ratings) > 0:
        average = sum(ratings) / len(ratings)
        print("Ratings:", ratings)
        print("Average Rating:", round(average, 2))
    else:
        print("No valid ratings available")
```
