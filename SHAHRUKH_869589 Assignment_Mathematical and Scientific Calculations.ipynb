## 7) Compound Interest: 
Write a Python program to calculate compound interest. Use the formula: A=P*(1+r/n)**n-t Where: 
A= total amount 
P = principal amount 
r = annual interest rate (decimal) 
n = number of times interest is compounded per year 
t = time in years 
Take P, r, n, and t as inputs from the user. 

```python
P = 10000
r = 0.05
n = 4
t = 2
```

```python
A = P * (1 + r / n) ** (n - t)
```

```python
print ('Total Amount = ', A)
```

```python
compound_interest = A - P
```

```python
print ('Compound Interest = ',compound_interest)
```
